# NUMPY
NumPy stands for Numerical Python and is used for handling large, multi-dimensional arrays and matrices. Unlike Python's built-in lists NumPy arrays provide efficient storage and faster processing for numerical and scientific computations. It offers functions for linear algebra and random number generation making it important for data science and machine learning.

### Install numpy array
Before install our packages and we setup our virtual environment:

- creation = python -m venv venv
- activate = venv/Scripts/

Then, we install our package

 *pip install numpy*

 we install our pavkages with help of requirements.txt file in our evironements.

*pip install -r numpy_create_array\requirements.txt --index-url https://pypi.org/simple*

- pip install → use Python’s package installer.

- -r requirements.txt → read the file called requirements.txt (inside the folder numpy_create_array) and install all the Python packages listed in it.

- --index-url https://pypi.org/simple
 → tells pip to download those packages from the official Python package repository (PyPI).
